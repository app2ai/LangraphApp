### Create sequencial workflow for prompt chaining where you have to create a blog
#### START-> Generate outline -> Generate blog details -> END

```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
from langchain_core.prompts import PromptTemplate
```

```python
load_dotenv()
```

```python
chat_model = ChatOpenAI()
```

```python
#Define states
class BlogState(TypedDict):
    blog_topic: str
    blog_outline: str
    blog_details: str


# Define state graph for blog creation workflow
graph = StateGraph(BlogState)

# Define python nodes
def generate_outline(state: BlogState):
    topic = state['blog_topic']
    outline_prompt = f'Generate a blog outline for {topic}'
    outline = chat_model.invoke(outline_prompt).content
    state['blog_outline'] = outline
    return state

def generate_blog(state: BlogState):
    prompt = f"Generate a details blog on given {state['blog_outline']}"
    blog = chat_model.invoke(prompt).content
    state['blog_details'] = blog
    return state

# Define nodes 
graph.add_node('generate_outline', generate_outline)
graph.add_node('generate_blog', generate_blog)

# Define edges
graph.add_edge(START, 'generate_outline')
graph.add_edge('generate_outline', 'generate_blog')
graph.add_edge('generate_blog', END)

# Compile code
workflow = graph.compile()

# invoke input
res = workflow.invoke({'blog_topic':'Effect of AI on Indian stock market'})

print(res)
```
