```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
```

```python
# Define states for workflow
class BMIState(TypedDict):
    weight: float # In Kg
    height: float # In Meters
    bmi_value: float
```

```python
# Define workflow graph
bmi_graph = StateGraph(BMIState)
```

```python
# python node
def calculate_bmi(state: BMIState) -> BMIState:
    w = state['weight']
    h = state['height']
    bmi = w/(h**2)
    state['bmi_value'] = bmi
    return state
```

```python
# Define workflow nodes
bmi_graph.add_node('cal_bmi', calculate_bmi)
```

```python
# Define edges
bmi_graph.add_edge(START, 'cal_bmi')
bmi_graph.add_edge('cal_bmi', END)
```

```python
# Compile workflow graph
workflow = bmi_graph.compile()
```

```python
# Execute or Invoke workflow
# input_weight = float(input("Enter your weight in KG: "))
# input_height = float(input("Enter your height in meter: "))
```

```python
initial_state = {'weight': 72.5, 'height': 1.71}

final_state = workflow.invoke(initial_state)

print(final_state)
```
